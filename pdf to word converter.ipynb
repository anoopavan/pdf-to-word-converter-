```python
import tkinter as tk
from tkinter import filedialog, messagebox
from pdf2docx import Converter
import os

def convert_pdf_to_word():
    pdf_path = filedialog.askopenfilename(title="Select PDF file", filetypes=[("PDF Files", "*.pdf")])
    if not pdf_path:
        return

    output_path = filedialog.asksaveasfilename(
        defaultextension=".docx",
        filetypes=[("Word Documents", "*.docx")],
        title="Save Word file as"
    )
    if not output_path:
        return

    try:
        cv = Converter(pdf_path)
        cv.convert(output_path, start=0, end=None)
        cv.close()
        messagebox.showinfo("Success", f"Conversion successful!\nSaved at:\n{output_path}")
    except Exception as e:
        messagebox.showerror("Error", f"Failed to convert:\n{str(e)}")

def run_ui():
    root = tk.Tk()
    root.title("PDF to Word Converter")
    root.geometry("300x150")

    label = tk.Label(root, text="Convert PDF to Word", font=("Arial", 14))
    label.pack(pady=20)

    btn = tk.Button(root, text="Select PDF and Convert", command=convert_pdf_to_word)
    btn.pack(pady=10)

    root.mainloop()
run_ui()
```
